# loc and iloc
loc is used for slecting rows or columns based on labels(index) of them<br>
iloc is used selection rows and columns based on integer position of them

```python
df = pd.DataFrame({
    'Name' : ['Ninja_1','Ninja_2','Ninja_3','Ninja_4'],
    'Department' : ['Marketing','Sales','Marketing','Engineering'],
    'Salary' : [50000,60000,55000,70000]

})
print(df)                                             
```

the index of Name is 0, index of Department is 1, index of Salary is 2

```python
# Selecting A Single 
df.loc[2]
```

```python
# Selecting Multiple Rows
df.loc[[1, 2, 3]]
# As you can see, the rows with labels '1','2','3' are selected. Here it is indicating towards the second, third and fourth rows in DataFrame since Python uses 0-based indexing technique.
```

```python
# Selecting Multiple Columns
df.loc[:, ['Name', 'Salary']]
```

```python
# Selecting A Subset Of Rows And Columns
df.loc[1:2, ['Name', 'Salary']]
```

```python
# updating the values of the dataframe using loc method
df.loc[1, 'Salary'] = 65000
print(df)
```

```python
df = pd.read_csv('student_performance.csv')
print(df.head())
```

```python
# we can select particular column from the dataframe using column name
# df.loc[row_labels, column_labels]
print(df.loc[:, 'Math_Score'])
```

```python
df.loc[:, ["Math_Score", "Science_Score"]]
```

```python
# updating values
df.loc[1, 'Math_Score'] = 95
print(df)
# so if you see the output the value of maths score is changed from ₹88 to 95 for the student with index 1. This is how we can update the values in a dataframe using loc method.
```

```python
# here we are selecting columns with condition 
df = pd.read_csv('student_performance.csv')
# here we are selecting the rows where the math score is less than or equal to 45 and then we are selecting the science score and english score for those rows
df.loc[df["Math_Score"] <= 45, ["Science_Score", "English_Score"]]
```

# iloc

```python

df = pd.DataFrame({
    'Name' : ['Ninja_1','Ninja_2','Ninja_3','Ninja_4'],
    'Department' : ['Marketing','Sales','Marketing','Engineering'],
    'Salary' : [50000,60000,55000,70000]

})
print(df)
```

```python
# Selecting Specific Rows And Columns
# df.iloc[row_positions, column_positions]
row = df.iloc[[1,3], [0,2]]
print(row)
```

```python
# Updating Values

df.iloc[1,2] = 100
print(df)
```

```python
# slicing the dataframe using iloc method
df.iloc[0:3, 0:2]
```

```python
import pandas as pd
df = pd.read_csv('student_performance.csv')
print(df.head())
```

```python
# Selecting Specific Rows And Columns
df.iloc[[1,3], [0,2]]
# here we are selecting the second and fourth row and first and third column of the dataframe using iloc method. The output will be a new dataframe containing the selected rows and columns.
```

```python
# df.iloc[row_positions, column_positions]
print(df.iloc[0:5, 0:3])
```

```python
# updating values
df.iloc[1, 1] = 95
print(df)
# so if you see the output the value of age is changed from 88 to 95 for the student with index 1. This is how we can update the values in a dataframe using iloc method.
```
